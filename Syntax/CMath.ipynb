- ### Trigonometric

```python
# Setup for oneline command %%cpp
import os, tempfile, subprocess
from IPython.core.magic import register_cell_magic
import shlex

@register_cell_magic
def cpp(line, cell):
    """
    Usage:
    %%cpp -i "input for cin" -- arg1 arg2 ...
    """
    tokens = shlex.split(line)
    input_data = None
    run_args = []

    # Parse stdin input
    if "-i" in tokens:
        idx = tokens.index("-i")
        if idx + 1 < len(tokens):
            input_data = tokens[idx + 1]

    # Parse program arguments after --
    if "--" in tokens:
        idx = tokens.index("--")
        run_args = tokens[idx + 1:]

    # Write temp C++ file
    with tempfile.NamedTemporaryFile(suffix=".cpp", delete=False, mode="w") as tmp_cpp:
        tmp_cpp.write(cell)
        cpp_path = tmp_cpp.name
    exe_path = cpp_path[:-4] + ".exe"

    try:
        # Compile
        compile_proc = subprocess.run(
            ["g++", "-std=c++23", "-O2", "-Wall", cpp_path, "-o", exe_path],
            capture_output=True,
            text=True
        )
        if compile_proc.returncode != 0:
            print("❌ Compilation failed:\n", compile_proc.stderr)
            return

        # Run program
        run_proc = subprocess.run(
            [exe_path] + run_args,
            input=input_data,      # feed stdin here
            capture_output=True,
            text=True
        )
        if run_proc.stdout:
            print(run_proc.stdout, end="")
        if run_proc.stderr:
            print("⚠️ Runtime error:\n", run_proc.stderr)

    finally:
        for f in (cpp_path, exe_path):
            try: os.remove(f)
            except: pass
```

**sin & cos & tan**

```python
%%cpp
#include <iostream>
#include <cmath>
#include <vector>
#include <numbers>
using namespace std;
using namespace std::numbers;

int main() {
    vector<double> nums = {pi/6, pi/4, pi/3, pi/2, pi};
    vector<double> sin_nums;
    vector<double> cos_nums;
    vector<double> tan_nums;
    for (auto& n : nums) {
        cout << n << " ";
    }
    cout << endl << "sin" << endl;
    for (auto& n : nums) {
        cout << sin(n) << " ";
        sin_nums.push_back(sin(n));
    }
    cout << endl << "cos" << endl;
    for (auto& n : nums) {
        cout << cos(n) << " ";
        cos_nums.push_back(cos(n));
    }
    cout << endl << "tan" << endl;
    for (auto& n : nums) {
        cout << tan(n) << " ";
        tan_nums.push_back(tan(n));
    }
    cout << endl;
}
```

**asin & acos & atan**

```python
%%cpp
#include <iostream>
#include <cmath>
#include <vector>
#include <numbers>
using namespace std;
using namespace std::numbers;

int main() {
    vector<double> nums = {0,0.25,0.5,0.75,1};
    vector<double> arcsin_nums;
    vector<double> arccos_nums;
    vector<double> arctan_nums;
    for (auto& n : nums) {
        cout << n << " ";
    }
    cout << endl << "arcsin" << endl;
    for (auto& n : nums) {
        cout << asin(n) << " ";
        arcsin_nums.push_back(asin(n));
    }
    cout << endl << "arccos" << endl;
    for (auto& n : nums) {
        cout << acos(n) << " ";
        arccos_nums.push_back(acos(n));
    }
    cout << endl << "arctan" << endl;
    for (auto& n : nums) {
        cout << atan(n) << " ";
        arctan_nums.push_back(atan(n));
    }
    cout << endl;
}
```

- ### Power

**power**

```python
%%cpp
#include <iostream>
#include <cmath>
#include <vector>
#include <numbers>
using namespace std;
using namespace std::numbers;

int main() {
    vector<double> nums = {0,1,e,pi};
    vector<double> exp_nums;
    vector<double> log_nums;
    vector<double> pow_nums;

    for (auto& n : nums) {
        cout << "exp(" << n << ") = " << exp(n) << endl;
        exp_nums.push_back(exp(n));
    }
    cout << endl;
    for (auto& n : nums) {
        cout << "log(" << n << ") = " << log(n) << endl;
        log_nums.push_back(log(n));
    }
    cout << endl;
    for (auto& n : nums) {
        cout << "pow(" << n << ", 2) = " << pow(n, 2) << endl;
        pow_nums.push_back(pow(n, 2));
    }
    cout << endl;
}
```
