- ### Basic type

```python
# Setup for oneline command %%cpp
import os, tempfile, subprocess
from IPython.core.magic import register_cell_magic
import shlex

@register_cell_magic
def cpp(line, cell):
    """
    Usage:
    %%cpp -i "input for cin" -- arg1 arg2 ...
    """
    tokens = shlex.split(line)
    input_data = None
    run_args = []

    # Parse stdin input
    if "-i" in tokens:
        idx = tokens.index("-i")
        if idx + 1 < len(tokens):
            input_data = tokens[idx + 1]

    # Parse program arguments after --
    if "--" in tokens:
        idx = tokens.index("--")
        run_args = tokens[idx + 1:]

    # Write temp C++ file
    with tempfile.NamedTemporaryFile(suffix=".cpp", delete=False, mode="w") as tmp_cpp:
        tmp_cpp.write(cell)
        cpp_path = tmp_cpp.name
    exe_path = cpp_path[:-4] + ".exe"

    try:
        # Compile
        compile_proc = subprocess.run(
            ["g++", "-std=c++23", "-O2", "-Wall", cpp_path, "-o", exe_path],
            capture_output=True,
            text=True
        )
        if compile_proc.returncode != 0:
            print("❌ Compilation failed:\n", compile_proc.stderr)
            return

        # Run program
        run_proc = subprocess.run(
            [exe_path] + run_args,
            input=input_data,      # feed stdin here
            capture_output=True,
            text=True
        )
        if run_proc.stdout:
            print(run_proc.stdout, end="")
        if run_proc.stderr:
            print("⚠️ Runtime error:\n", run_proc.stderr)

    finally:
        for f in (cpp_path, exe_path):
            try: os.remove(f)
            except: pass
```

**array: data_type arr[n]**

```python
%%cpp
#include <iostream>
using namespace std;
int main() {
    int arr[5] = {1,2,3,4,5};
    for (int x : arr){
        cout << x << " ";
    }
}
```

**pointer: data_type * ptr**

```python
%%cpp
#include <iostream>
using namespace std;

int main() {
    int a = 5;
    int* p = &a; // * for value at address, & for address of variable
    *p = 10; // change value at address
    cout << a << '\n' << *p << '\n' << p << endl;
    }
```

```python
%%cpp
#include <iostream>
using namespace std;

int main() {
    string str = "Hello world";
    string *p = &str;
    string **pp = &p;
    cout << p << '\n' << pp << '\n' << **pp << endl;
    }
```

```python
%%cpp
#include <iostream>
#include <vector>
using namespace std;

int main() {
    vector<int> vec = {1,2,3,4,5};
    int *p = &vec[0];
    while (p != &vec[5]) {
        cout << *p << " ";
        p++;
    }
    cout << endl;
    }
```

**null pointer**

```python
%%cpp
#include <iostream>
#include <vector>
using namespace std;

int main() {
    vector<int> vec = {1,2,3,4,5};
    int * p = nullptr; // null pointer
    for (int x : vec) {
        p = &x;
        *p += 2; // change value at address
        cout << x << " ";
    }
    cout << endl;
    }
```

**reference**

```python
%%cpp
#include <iostream>
using namespace std;

int main() {
    int a = 5;
    int & r = a; // reference
    r = 10; // change value via reference
    cout << a << '\n' << r << endl;
    }
```

```python
%%cpp
#include <iostream>
using namespace std;

int main() {
int x = 10;

// pointer
int* p = &x;     // here '&x' means “address of x”
// reference
int& r = x;      // here '&' means “reference to x”

*p += 5;    // change x by pointer
cout << *p << " " << r << endl;

r += 5;   // change x by reference
cout << *p << " " << r << endl;
}
```

**std::string**

```python
%%cpp
#include <iostream>
using namespace std;

int main() {
    string PI = "Hello world";
    for (char c : PI) {
        cout << c << ' ';
    }
    cout << endl;
    }
```

**std::vector<int>**

```python
%%cpp
#include <iostream>
#include <vector>
#include <string>
using namespace std;

int main() {
    vector<string> s = {"Hello", "world"};
    vector<float> v = {3.14, 2.71};
    for (float x : v) {
        cout << x << ' ';
    }
    cout << endl;
    for (int i = 0; i < s.size(); i++) {
        cout << s[i] << ' ';
    }
    cout << endl;
    }
```

**std::set<int>**

```python
%%cpp
#include <iostream>
#include <set>
using namespace std;

int main() {
    set<int> s;
    s.insert(1);
    s.insert(2);
    s.insert(3);
    s.insert(2);
    s.erase(3); 
    for (int x : s) {
        cout << x << " ";
    }
}
```

**std::tuple<int>**

```python
%%cpp
#include <iostream>
#include <tuple>
#include <string>
using namespace std;

int main() {
    tuple<int, string, double> t = {1, "hello", 3.14};

    // Lambda to print all elements of the tuple
    auto print = [](const auto&... elems) {
        ((cout << elems << " "), ...);
    };

    apply(print, t);   // Unpacks tuple into lambda
    cout << endl;
}
```

**std::array** 

```python
%%cpp
#include <iostream>
#include <array>
using namespace std;

int main() {
    array<int, 5> arr = {1, 2, 3, 4, 5};
    for (int x : arr) {
        cout << x << ' ';
    }
    cout << endl;
}
```

**std::map**

```python
%%cpp
#include <iostream>
#include <map>
using namespace std;

int main() {
    map<string, int> m;
    m["one"] = 1;
    m["two"] = 2;
    m["three"] = 3;
    for (const auto& [key, value]: m){
        cout << key << ": " << value << endl;
    }
    }
```

**std::list**

```python
%%cpp
#include <iostream>
#include <list>
using namespace std;

int main() {
    list<int> l={1,2,3,4,5};
    cout << "Size of l: " << l.size() << endl;
    l.push_back(6);
    l.push_front(0);
    for (auto &x : l){
        cout << x << " ";
    }
    cout << endl;
    l.pop_back();
    l.pop_front();
    for (auto &x : l){
        cout << x << " ";
    }
    cout << endl;
    }
```

- ### Type aliases

using A = [modifier] data_type

```python
%%cpp
#include <iostream>
#include <vector>
using namespace std;
using vector_string = vector<string>;
using vector_int = vector<int>;

int main() {
    vector_string vs = {"Hello", "world"};
    vector_int vi = {1, 2, 3};
    for (const auto& s : vs) {
        cout << s << ' ';
    }
    cout << endl;
    for (auto it = vi.begin(); it != vi.end(); ++it) {
        cout << *it << ' '; // dereference iterator
    }
    cout << endl;
    }
```

**text replacement**

```python
%%cpp
#include <iostream>
using namespace std;
#define MAX(a,b) ((a) > (b) ? (a) : (b))

int main() {
    int x = 10, y = 20;
    cout << "Max of " << x << " and " << y << " is " << MAX(x, y) << endl;
    return 0;
}
```
