- ### Self-defined function

**calling function**

````python
%%cpp
#include <iostream>
using namespace std;

void print(const string& msg); // Forward declaration
// prefix "extern" is needed if call from another file

int main() {    // Main function
    print("Hello from printf!");
}

void print(const string& msg) { // Function definition
    printf("%s\n",msg.c_str());
}
````

**default input**

````python
%%cpp
#include <iostream>
using namespace std;

int minmum(int a = 0, int b = 0) { // Default arguments
    return (a < b) ? a : b;
}
int main() {    // Main function
    cout << minmum(3, 5) << endl; 
    cout << minmum(3) << endl;    
    cout << minmum() << endl;
}

````

- ### Lambda function

```[capture](parameters) -> return_type {function_body}```

**capture nothing - []**

````python
%%cpp
#include <iostream>
using namespace std;

int main() {
    cout << [](int x, int y){return x + y;}(3, 5) << endl; // Lambda function
}
````

````python
%%cpp
#include <iostream>
using namespace std;

int main() {    
    auto mean = [](double x, double y, double z) {
        return (x + y + z) / 3;
    };
    cout << mean(3, 5, 6) << endl;
}
````

**capture local variables x, y- [x, y]**

````python
%%cpp
#include <iostream>
#include <string>
using namespace std;

int main() {
    string x = "Hello";
    string y = "World";
    auto join = [x, y](string delimiter = ",") { cout << x + delimiter + y << endl; };
    join(" - ");// Capture outside variables
    join();    // Use default argument
}
````

**capture all local variables by value- [=]**

````python
%%cpp
#include <iostream>
using namespace std;

int main() {
    int n = 5;
    auto acc = [=](int i, auto self) -> int {
        return i <= 1? 1 : i + self(i-1, self);
    };
    cout << acc(n, acc) << endl;
}
````

**capture all local variables by reference- [&]**

````python
%%cpp
#include <iostream>
using namespace std;

int main() {
    int n = 5;
    auto inc = [&n]() { ++n; }; 
    inc(); 
    cout << n << endl;
}
````

**vector component-wise operation**

````python
%%cpp
#include <iostream>
#include <vector>
#include <algorithm>
using namespace std;

int main() {
    vector<int> v = {1, 2, 3, 4, 5};
    for_each(v.begin(), v.end(), [](int &x){ x *=x ; }); // Modify in place
    for (const auto &x : v) cout << x << " "; cout << endl;
}
````

**customize algorithm**

````python
%%cpp
#include <iostream>
#include <vector>
#include <algorithm>
using namespace std;

int main() {
    vector<int> v = {5, 3, 2, 8, 1};
    sort(v.begin(), v.end(), [](int a, int b) { return a > b; }); // Custom sort
    for (int x : v) cout << x << " ";
    cout << endl;
}
````

**self-reference**

````python
%%cpp
#include <iostream>
using namespace std;

int main() {
    auto factorial = [] (auto self, int n) -> int { // Lambda function
        return (n <= 1) ? 1 : n * self(self,n-1); // self-reference
    };
    cout << factorial(factorial, 5) << endl;
} 
````
