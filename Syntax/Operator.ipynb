- ### Assignment operators

```python
# Setup for oneline command %%cpp
import os, tempfile, subprocess
from IPython.core.magic import register_cell_magic
import shlex

@register_cell_magic
def cpp(line, cell):
    """
    Usage:
    %%cpp -i "input for cin" -- arg1 arg2 ...
    """
    tokens = shlex.split(line)
    input_data = None
    run_args = []

    # Parse stdin input
    if "-i" in tokens:
        idx = tokens.index("-i")
        if idx + 1 < len(tokens):
            input_data = tokens[idx + 1]

    # Parse program arguments after --
    if "--" in tokens:
        idx = tokens.index("--")
        run_args = tokens[idx + 1:]

    # Write temp C++ file
    with tempfile.NamedTemporaryFile(suffix=".cpp", delete=False, mode="w") as tmp_cpp:
        tmp_cpp.write(cell)
        cpp_path = tmp_cpp.name
    exe_path = cpp_path[:-4] + ".exe"

    try:
        # Compile
        compile_proc = subprocess.run(
            ["g++", "-std=c++23", "-O2", "-Wall", cpp_path, "-o", exe_path],
            capture_output=True,
            text=True
        )
        if compile_proc.returncode != 0:
            print("❌ Compilation failed:\n", compile_proc.stderr)
            return

        # Run program
        run_proc = subprocess.run(
            [exe_path] + run_args,
            input=input_data,      # feed stdin here
            capture_output=True,
            text=True
        )
        if run_proc.stdout:
            print(run_proc.stdout, end="")
        if run_proc.stderr:
            print("⚠️ Runtime error:\n", run_proc.stderr)

    finally:
        for f in (cpp_path, exe_path):
            try: os.remove(f)
            except: pass
```

```python
%%cpp
#include <iostream>
using namespace std;

int main() {
    int a = 5;
    int b = 10;
    a += b; // a = a + b
    cout << a << endl;
    a -= b; // a = a - b
    cout << a << endl;

    a *= b; // a = a * b
    cout << a << endl;
    a /= b; // a = a / b
    cout << a << endl;
    
    a %= b; // a = a % b
    cout << a << endl;
}
```

- ### Arithmetic operators

```python
%%cpp
#include <iostream>
using namespace std;
int main() {
    double pi = 3.1415;
    double e = 2.7182;
    double sum = pi + e;
    double sub = pi - e;
    double mul = pi * e;
    double div = pi / e;
    double mod = (int)pi % (int)e; // cast to int for modulus
    double preinc = ++pi; // pre-increment
    double postinc = pi++; // post-increment
    double predec = --e;  // pre-decrement
    double postdec = e--; // post-decrement
    cout << sum << ' ' << sub << endl;
    cout << mul << ' ' << div << endl;
    cout << mod << endl;
    cout << preinc << ' ' << postinc << endl;
    cout << predec << ' ' << postdec << endl;
}
```

integer division

```python
%%cpp
#include <iostream>
using namespace std;
int main() {
    int a = 2;
    int b = a++; // post-increment
    double c = a/b; // integer division
    double d = (double)a/(double)b; // type casting
    cout << a << ' ' << b << endl;
    cout << c << ' ' << d << endl;
}
```

- ### Bitwise operators

```python
%%cpp
#include <iostream>
using namespace std;

int main() {
    int a = 5; // 5 = 0000 0101 in binary
    int b = 10; // 10 = 0000 1010 in binary
    cout << (a & b) << ' ' << (a | b) << ' ' << (a ^ b) << endl; // bitwise AND, OR, XOR
    // a & b = 0000 0000 = 0
    // a | b = 0000 1111 = 15
    // a ^ b = 0000 1111 = 15
    cout << (~a) << ' ' << (~b) << endl; // bitwise NOT
    // ~a = 1111 1010 = -6 (two's complement) ~x = -(x+1)
    // ~b = 1111 0101 = -11 (two's complement)
    cout << (a << 1) << ' ' << (b >> 1) << endl; // left shift, right shift
    // a << 1 = 0000 1010 = 10
    // b >> 1 = 0000 0101 = 5
}
```

**binary**

```python
%%cpp
#include <iostream>
using namespace std;

int main() {
    int n = 13;
    for (int i = 7; i >= 0; --i) {   // print 8 bits
        cout << ((n >> i) & 1);
    }
    cout << endl;
}
```
