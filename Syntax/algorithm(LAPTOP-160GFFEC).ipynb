- ### Search

````python
# Setup for oneline command %%cpp
import os, tempfile, subprocess
from IPython.core.magic import register_cell_magic # type: ignore
import shlex

@register_cell_magic
def cpp(line, cell):
    """
    Usage:
    %%cpp -i "input for cin" -- arg1 arg2 ...
    """
    tokens = shlex.split(line)
    input_data = None
    run_args = []

    # Parse stdin input
    if "-i" in tokens:
        idx = tokens.index("-i")
        if idx + 1 < len(tokens):
            input_data = tokens[idx + 1]

    # Parse program arguments after --
    if "--" in tokens:
        idx = tokens.index("--")
        run_args = tokens[idx + 1:]

    # Write temp C++ file
    with tempfile.NamedTemporaryFile(suffix=".cpp", delete=False, mode="w") as tmp_cpp:
        tmp_cpp.write(cell)
        cpp_path = tmp_cpp.name
    exe_path = cpp_path[:-4] + ".exe"

    try:
        # Compile
        compile_proc = subprocess.run(
            ["g++", "-std=c++23", "-O2", "-Wall", cpp_path, "-o", exe_path],
            capture_output=True,
            text=True
        )
        if compile_proc.returncode != 0:
            print("❌ Compilation failed:\n", compile_proc.stderr)
            return

        # Run program
        run_proc = subprocess.run(
            [exe_path] + run_args,
            input=input_data,      # feed stdin here
            capture_output=True,
            text=True
        )
        if run_proc.stdout:
            print(run_proc.stdout, end="")
        if run_proc.stderr:
            print("⚠️ Runtime error:\n", run_proc.stderr)

    finally:
        for f in (cpp_path, exe_path):
            try: os.remove(f)
            except: pass
````

**find**

````python
%%cpp
#include <iostream>
#include <algorithm>
#include <vector>
using namespace std;

void reply(auto it, vector<int> &v) {
        if (it != v.end()) {
        cout << "Found 5 at position: " << distance(v.begin(), it) << endl;
    } else {
        cout << "5 not found" << endl;
    }
    };

int main() {
    vector<int> v = {5, 2, 9, 1, 5, 6};
    auto it1 = find(v.begin(), v.end(), 5); // return iterator
    auto it2 = find(v.begin(), v.end(), 10); // return end iterator
    reply(it1,v);
    reply(it2,v);
}
````

````python
%%cpp
#include <iostream>
#include <algorithm>
#include <vector>
using namespace std;

void reply(auto it, vector<int> &v) {
        if (it != v.end()) {
        cout << "Found first even element at position: " << distance(v.begin(), it) << endl;
    } else {
        cout << "not found" << endl;
    }
    };

int main() {
    vector<int> v = {5, 2, 9, 1, 5, 6};
    auto it = find_if(v.begin(), v.end(), [](int x){ return x % 2 == 0; }); // find first even number
    reply(it,v);
}
````

- ### Sort

**sort**

````python
%%cpp
#include <iostream>
#include <algorithm>
#include <vector>
using namespace std;

int main() {
    vector<int> v = {1, 6, 5, 3, 3, 5};
    sort(v.begin(), v.end());
    for (auto x : v){
        cout << x << " ";
    }
    cout << endl;
}
````

````python
%%cpp
#include <iostream>
#include <algorithm>
#include <vector>
using namespace std;

int main() {
    vector<int> v = {1, 6, 5, 3, 3, 5};
    sort(v.begin(), v.end(), [v] (int x, int y) {return x < y;});
    for (auto x : v){
        cout << x << " ";
    }
    cout << endl;
}
````

- #### Modification

**reverse**

````python
%%cpp
#include <iostream>
#include <algorithm>
#include <vector>
using namespace std;

int main() {
    vector<double> v = {1,5,3};
    reverse(v.begin(), v.end());
    for (auto x : v){
        cout << x << " ";
    }
    cout << endl;
}
````

**shuffle**

````python
%%cpp
#include <iostream>
#include <algorithm>
#include <vector>
#include <random>
using namespace std;

int main() {
    vector<int> v = {1,3,4,2};
    random_device rd; // create random number generator;
    mt19937 g(rd());
    shuffle(v.begin(), v.end(), g);
    for (auto x : v){
        cout << x << " ";
    }
    cout << endl;
}
````

**rotation**

````python
%%cpp
#include <iostream>
#include <algorithm>
#include <vector>
using namespace std;

int main() {
    vector<int> v = {1,3,4,2};
    auto it = v.begin();
    advance(it, 2);
    rotate(v.begin(), it, v.end());
    for (auto x : v){
        cout << x << " ";
    }
    cout << endl;
}
````

**unique**

````python
%%cpp
#include <iostream>
#include <algorithm>
#include <vector>
using namespace std;

int main() {
    vector<double> v = {1,5,3,3};
    auto it = unique(v.begin(), v.end());
    for (auto i = v.begin(); i< it; i++){
        cout << *i << " ";
    }
    cout << endl;
}
````

- ### Moving

**copy**

````python
%%cpp
#include <iostream>
#include <algorithm>
#include <vector>
#include <iterator> // using for back_inserter
using namespace std;

int main() {
    vector<int> v = {1,3,4,2};
    vector<int> u;
    copy(v.begin(), v.end(), back_inserter(u));
    for (auto x : u){
        cout << x << " ";
    }
}
````

**copy_if**

````python
%%cpp
#include <iostream>
#include <algorithm>
#include <vector>
#include <iterator> // using for back_inserter
using namespace std;

int main() {
    vector<int> v = {1,3,4,2};
    vector<int> u;
    copy_if(v.begin(), v.end(), back_inserter(u), [](int x){return x % 2 == 0;});
    for (auto x : u){
        cout << x << " ";
    }
}
````

**move**

````python
%%cpp
#include <iostream>
#include <algorithm>
#include <vector>
using namespace std;

int main() {
    vector<int> v = {1,3,4,2};
    vector<int> u;
    u = move(v);
    for (auto x : u){
        cout << x << " ";
    }
    cout << "\nsize of v: " << v.size() << endl; // v is now empty
}
````

**swap**

````python
%%cpp
#include <iostream>
#include <algorithm>
#include <vector>
using namespace std;

int main() {
    vector<int> v = {1,3,4,2};
    vector<int> u = {3,1,2,4};
    swap(u, v);
    for (auto x : u){
        cout << x << " ";
    }
    cout << endl;
    for (auto x : v){
        cout << x << " ";
    }
}
````

- ### Set operations (on sorted containers)

**set_union**

````python
%%cpp
#include <iostream>
#include <algorithm>
#include <set>
#include <iterator> // using for set_union
using namespace std;

int main() {
    set<int> v = {1,3,4,2};
    set<int> u = {6,1,5,0};
    set<int> w;
    set_union(u.begin(), u.end(), v.begin(), v.end(), inserter(w, w.begin()));
    for (auto x : w){
        cout << x << " ";
    }
}
````

**set_intersection**

````python
%%cpp
#include <iostream>
#include <algorithm>
#include <vector>
#include <iterator> // using for set_intersection
using namespace std;

int main() {
    vector<int> v = {1,3,4,2};
    vector<int> u = {6,1,5,0};
    vector<int> w;
    
    sort(v.begin(), v.end()); // sort is needed for set operations
    sort(u.begin(), u.end());
    set_intersection(u.begin(), u.end(), v.begin(), v.end(), inserter(w, w.begin()));
    for (auto x : w){
        cout << x << " ";
    }
}
````

**set_difference**

````python
%%cpp
#include <iostream>
#include <algorithm>
#include <map>
#include <iterator> // using for set_intersection
using namespace std;

int main() {
    map<int, int> v = {{1,3},{4,2}};
    map<int, int> u = {{1,3},{2,3}};
    map<int, int> w;
    
    set_intersection(u.begin(), u.end(), v.begin(), v.end(), inserter(w, w.begin()));
    for (auto [k, v] : w){
        cout << k << " " << v << " ";
    }
}
````

**set_symmetric_difference**

````python
%%cpp
#include <iostream>
#include <algorithm>
#include <map>
#include <iterator> // using for set_intersection
using namespace std;

int main() {
    map<int, int> v = {{1,3},{4,2}};
    map<int, int> u = {{1,3},{2,3}};
    map<int, int> w;
    
    set_symmetric_difference(u.begin(), u.end(), v.begin(), v.end(), inserter(w, w.begin()));
    for (auto [k, v] : w){
        cout << k << " " << v << " ";
    }
}
````
