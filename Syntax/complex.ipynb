- ### Real & Image

```python
# Setup for oneline command %%cpp
import os, tempfile, subprocess
from IPython.core.magic import register_cell_magic # type: ignore
import shlex

@register_cell_magic
def cpp(line, cell):
    """
    Usage:
    %%cpp -i "input for cin" -- arg1 arg2 ...
    """
    tokens = shlex.split(line)
    input_data = None
    run_args = []

    # Parse stdin input
    if "-i" in tokens:
        idx = tokens.index("-i")
        if idx + 1 < len(tokens):
            input_data = tokens[idx + 1]

    # Parse program arguments after --
    if "--" in tokens:
        idx = tokens.index("--")
        run_args = tokens[idx + 1:]

    # Write temp C++ file
    with tempfile.NamedTemporaryFile(suffix=".cpp", delete=False, mode="w") as tmp_cpp:
        tmp_cpp.write(cell)
        cpp_path = tmp_cpp.name
    exe_path = cpp_path[:-4] + ".exe"

    try:
        # Compile
        compile_proc = subprocess.run(
            ["g++", "-std=c++23", "-O2", "-Wall", cpp_path, "-o", exe_path],
            capture_output=True,
            text=True
        )
        if compile_proc.returncode != 0:
            print("❌ Compilation failed:\n", compile_proc.stderr)
            return

        # Run program
        run_proc = subprocess.run(
            [exe_path] + run_args,
            input=input_data,      # feed stdin here
            capture_output=True,
            text=True
        )
        if run_proc.stdout:
            print(run_proc.stdout, end="")
        if run_proc.stderr:
            print("⚠️ Runtime error:\n", run_proc.stderr)

    finally:
        for f in (cpp_path, exe_path):
            try: os.remove(f)
            except: pass
```

```python
%%cpp
#include <iostream>
#include <complex>
using namespace std;

int main() {
    complex<double> z(1.0, 2.0);
    cout << z.real() << " " << z.imag() << endl;
}
```

- ### Operation

**basics**

```python
%%cpp
#include <iostream>
#include <complex>
using namespace std;

int main() {
    complex<double> z1(1.0, 2.0);
    complex<double> z2(2.0, -2.0);
    complex<double> z3 = z1 + z2;
    complex<double> z4 = z1 - z2;
    complex<double> z5 = z1 * z2;
    complex<double> z6 = z1 / z2;
    cout << z3 << "\n" << z4 << endl;
    cout << z5 << "\n" << z6 << endl;
}
```

**polar**

```python
%%cpp
#include <iostream>
#include <complex>
using namespace std;

int main() {
    complex<double> z(1.0, 2.0);
    complex<double> z1 = conj(z);
    double norm = abs(z);
    double arguement = arg(z);
    cout << z1 << endl;
    cout << norm << "\n" << arguement << endl;
}
```

**trigonometric**

```python
%%cpp
#include <iostream>
#include <complex>
using namespace std;

int main() {
    complex<double> z(1.0, 2.0);
    complex<double> z1 = exp(z);
    complex<double> z2 = sin(z);
    complex<double> z3 = cos(z);
    cout << z1 << endl;
    cout << z2 << endl;
    cout << z3 << endl;
}
```

- ### Polar

```python
%%cpp
#include <iostream>
#include <complex>
#include <numbers>
using namespace std;
using namespace std::numbers;

int main() {
    double r = 3;
    double theta = pi / 4;
    complex z = polar(r, theta);
    cout << z << endl;
}
```
