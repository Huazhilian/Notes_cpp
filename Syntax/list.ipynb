- ### Merge

**merging**

```python
%%cpp
#include <iostream>
#include <list>
using namespace std;

int main() {
    list<int> l1={-6,1,2,3,4,5};
    list<int> l2={0,-1,-2,-3,-4,-5};
    auto it = next(l1.begin(), 1);
    l2.reverse();
    l1.splice(it, l2);
    cout << "l1 after splice:\n";
    for (auto &x : l1){
        cout << x << " ";
    };
    cout << "\nl2 after splice: \n(empty)";
    for (auto &x : l2){
        cout << x << " ";
    };
    cout << endl;
    }
```

**splicing**

```python
%%cpp
#include <iostream>
#include <list>
using namespace std;

int main() {
    list<int> l1={-6,1,2,3,4,5};
    list<int> l2={0,-1,-2,-3,-4,-5};
    auto it = next(l1.begin(), 1);
    l2.reverse();
    l1.splice(it, l2);
    cout << "l1 after splice:\n";
    for (auto &x : l1){
        cout << x << " ";
    };
    cout << "\nl2 after splice: \n(empty)";
    for (auto &x : l2){
        cout << x << " ";
    };
    cout << endl;
    }
```

```python
%%cpp
#include <iostream>
#include <list>
using namespace std;

int main() {
    list<int> l={3,2,1,5,4};
    l.sort();
    for (auto &x : l){
        cout << x << " ";
    }
    cout << endl;
    l.reverse();
    for (auto &x : l){
        cout << x << " ";
    }
    cout << endl;
    }
```
