```python
# Setup for oneline command %%cpp
import os, tempfile, subprocess
from IPython.core.magic import register_cell_magic
import shlex

@register_cell_magic
def cpp(line, cell):
    """
    Usage:
    %%cpp -i "input for cin" -- arg1 arg2 ...
    """
    tokens = shlex.split(line)
    input_data = None
    run_args = []

    # Parse stdin input
    if "-i" in tokens:
        idx = tokens.index("-i")
        if idx + 1 < len(tokens):
            input_data = tokens[idx + 1]

    # Parse program arguments after --
    if "--" in tokens:
        idx = tokens.index("--")
        run_args = tokens[idx + 1:]

    # Write temp C++ file
    with tempfile.NamedTemporaryFile(suffix=".cpp", delete=False, mode="w") as tmp_cpp:
        tmp_cpp.write(cell)
        cpp_path = tmp_cpp.name
    exe_path = cpp_path[:-4] + ".exe"

    try:
        # Compile
        compile_proc = subprocess.run(
            ["g++", "-std=c++23", "-O2", "-Wall", cpp_path, "-o", exe_path],
            capture_output=True,
            text=True
        )
        if compile_proc.returncode != 0:
            print("❌ Compilation failed:\n", compile_proc.stderr)
            return

        # Run program
        run_proc = subprocess.run(
            [exe_path] + run_args,
            input=input_data,      # feed stdin here
            capture_output=True,
            text=True
        )
        if run_proc.stdout:
            print(run_proc.stdout, end="")
        if run_proc.stderr:
            print("⚠️ Runtime error:\n", run_proc.stderr)

    finally:
        for f in (cpp_path, exe_path):
            try: os.remove(f)
            except: pass
```

- ### Modification

**insert/emplace**

```python
%%cpp
#include <iostream>
#include <map>
#include <string>
using namespace std;

int main() {
    map <string, string> uppercase;
    for (int i = 0; i < 26; i++)
    {
        char lower = 'a' + i;
        char upper = 'A' + i;
        uppercase.insert({string(1, lower),string(1, upper)});
    }
    for (auto [lower, upper] : uppercase)
    {
        cout << lower << " -> " << upper << endl;
    }
}
```

```python
%%cpp
#include <iostream>
#include <map>
#include <string>
using namespace std;

int main() {
    map <string, string> lowercase;
    for (int i = 0; i < 26; i++)
    {
        char lower = 'A' + i;
        char upper = 'a' + i;
        lowercase.emplace(string(1, lower),string(1, upper)); // C++11 feature
    }
    for (auto [upper, lower] : lowercase)
    {
        cout << upper << " -> " << lower << endl;
    }
}
```

**erase**

```python
%%cpp
#include <iostream>
#include <map>
#include <string>
using namespace std;

int main() {
    map <double, double> reciprocals;
    for (int i = 1; i <= 10; i++)
    {
        reciprocals[i] = 1.0 / i;
    }
    reciprocals.erase(10.0); // remove key 5.0

    for (auto it = reciprocals.begin(); it != reciprocals.end();(it != reciprocals.end())?++it:it)
    {
        auto next = it;
        ++next;  // move to next before erase
        reciprocals.erase(it);
        it = next;
    }
    for (auto [key, value] : reciprocals)
    {
        cout << "1/" << key << " = " << value << endl;
    }
    cout << "Size of map: " << reciprocals.size() << endl;
}
```

- ### Operation

**swap**

```python
%%cpp
#include <iostream>
#include <map>
using namespace std;

int main() {
    map<int, int> m1 = {{1, 2}, {3, 4}};
    map<int, int> m2 = {{2, 1}, {4, 3}};
    swap(m1, m2); // symmetric swap 
    for (auto [key, value] : m1) {
        cout << key << "->" << value << " ";
    }
    cout << endl;
    for (auto [key, value] : m2) {
        cout << key << "->" << value << " ";
    }
    cout << endl;

    m1.swap(m2); // member function swap
    for (auto [key, value] : m1) {
        cout << key << "->" << value << " ";
    }
    cout << endl;
    for (auto [key, value] : m2) {
        cout << key << "->" << value << " ";
    }
    cout << endl;
}
```

**slicing**

```python
%%cpp
#include <iostream>
#include <map>
using namespace std;

int main() {
map <int, int> squares;
    for (int i = 1; i <= 10; i++)
    {
        squares[i] = i*i;
    }
    int low = 3;
    int high = 7;
    auto start = squares.lower_bound(low); // lower_bound: first key >= low
    auto end = squares.upper_bound(high); // upper_bound: first key > high
    for (auto it = start; it != end; ++it)
    {
        cout << "Square of " << it->first << " is " << it->second << endl;
        // it->first is the key, it->second is the value
    }
}
```

**find**

```python
%%cpp
#include <iostream>
#include <map>
using namespace std;

int main() {
map <int, int> squares;
    for (int i = 1; i <= 10; i++)
    {
        squares[i] = i*i;
    }
    int key = 5;
    auto it = squares.find(key);
    if (it != squares.end())
    {
        cout << "Found: Square of " << it->first << " is " << it->second << endl;
    }
    else
    {
        cout << "Key " << key << " not found." << endl;
    }
}
```

- ### Function parameters

**reference (modifiable)**

```python
%%cpp
#include <iostream>
#include <map>
using namespace std;

void squared(map<int, int> &m)
    {
        for (auto it = m.begin(); it != m.end(); ++it)
        {
            m[it->first] = it->second * it->second;
            cout << "Square of " << it->first  << " is " << it->second << endl;
    }
    }

int main() {
    map<int, int> m;
    for (int i = 1; i <= 5; i++)
    {
        m[i] = i;
    }
    squared(m);
}
```

**const reference (read only)**

```python
%%cpp
#include <iostream>
#include <map>
using namespace std;

int sumvalue(const map<int, int> &m)
{
    int sum = 0;
    for (auto [k, v] : m)
    {
        sum += v;
    }
    return sum;
}

int main() {
    map<int, int> m;
    for (int i = 1; i <= 5; i++)
    {
        m[i] = i*i;
    }
    cout << sumvalue(m) << endl;
}
```

**value (copied)**

```python
%%cpp
#include <iostream>
#include <map>
using namespace std;

map<int, int> even(map<int, int> m)
{
    map<int, int> m_even;
    for (auto [k, v] : m)
    {
        if (k % 2 == 0)
        {
            m_even[k] = v;
        }
    }
    return m_even;
}

int main() {
    map<int, int> m;
    for (int i = 1; i <= 5; i++)
    {
        m[i] = i*i;
    }
    map<int, int> m_even = even(m);
    for (auto [k, v] : m_even)
    {
        cout << k << " -> " << v << endl;
    }
}
```
