- ### Convert string to number

```python
# Setup for oneline command %%cpp
import os, tempfile, subprocess
from IPython.core.magic import register_cell_magic # type: ignore
import shlex

@register_cell_magic
def cpp(line, cell):
    """
    Usage:
    %%cpp -i "input for cin" -- arg1 arg2 ...
    """
    tokens = shlex.split(line)
    input_data = None
    run_args = []

    # Parse stdin input
    if "-i" in tokens:
        idx = tokens.index("-i")
        if idx + 1 < len(tokens):
            input_data = tokens[idx + 1]

    # Parse program arguments after --
    if "--" in tokens:
        idx = tokens.index("--")
        run_args = tokens[idx + 1:]

    # Write temp C++ file
    with tempfile.NamedTemporaryFile(suffix=".cpp", delete=False, mode="w") as tmp_cpp:
        tmp_cpp.write(cell)
        cpp_path = tmp_cpp.name
    exe_path = cpp_path[:-4] + ".exe"

    try:
        # Compile
        compile_proc = subprocess.run(
            ["g++", "-std=c++23", "-O2", "-Wall", cpp_path, "-o", exe_path],
            capture_output=True,
            text=True
        )
        if compile_proc.returncode != 0:
            print("❌ Compilation failed:\n", compile_proc.stderr)
            return

        # Run program
        run_proc = subprocess.run(
            [exe_path] + run_args,
            input=input_data,      # feed stdin here
            capture_output=True,
            text=True
        )
        if run_proc.stdout:
            print(run_proc.stdout, end="")
        if run_proc.stderr:
            print("⚠️ Runtime error:\n", run_proc.stderr)

    finally:
        for f in (cpp_path, exe_path):
            try: os.remove(f)
            except: pass
```

**istringstream**

```python
%%cpp
#include <iostream>
#include <sstream>
#include <string>
using namespace std;

int main() {
    string s = "123";
    int x;
    istringstream(s) >> x;
    cout << x + 1 << "\n";
}
```

```python
%%cpp
#include <iostream>
#include <string>
using namespace std;

int main() {
    string s = "123";
    int x = stoi(s); // string to int
    double y = stod(s); // string to double
    cout << x + 1 << "\n";
    cout << y + 1 << "\n";
}
```

- ### Convert number to string

```python
%%cpp
#include <iostream>
#include <sstream>
#include <string>
using namespace std;

int main() {
    double pi = 3.14;
    ostringstream oss;
    oss << pi;
    string s = oss.str();
    cout << s + "!" << "\n";
}
```

```python
%%cpp
#include <iostream>
#include <string>
using namespace std;

int main() {
    double pi = 3.14;
    string s = to_string(pi);
    cout << s + "!" << "\n";
}
```

- ### Split a string

 **by whitespace**

```python
%%cpp
#include <iostream>
#include <sstream>
#include <string>
#include <vector>
using namespace std;

int main() {
    string s = "Hello, World!";
    istringstream iss(s);
    vector<string> words;
    string word;
    while (iss >> word) {
        words.push_back(word);
    }
    for (const string& w : words) {
        cout << w << "\n";
    }
}
```

**by custom delimiter**

```python
%%cpp
#include <iostream>
#include <sstream>
#include <string>
#include <vector>
using namespace std;

int main() {
    string s = "a,b,c,d";
    istringstream iss(s);
    vector<string> words;
    string token;
    while (getline(iss, token, ',')) {
        words.push_back(token);
    }
    for (const string& w : words) {
        cout << w << "\n";
    }
}
```
