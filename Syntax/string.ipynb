- ### Declaration


```python
# Setup for oneline command %%cpp
import os, tempfile, subprocess
from IPython.core.magic import register_cell_magic
import shlex

@register_cell_magic
def cpp(line, cell):
    """
    Usage:
    %%cpp -i "input for cin" -- arg1 arg2 ...
    """
    tokens = shlex.split(line)
    input_data = None
    run_args = []

    # Parse stdin input
    if "-i" in tokens:
        idx = tokens.index("-i")
        if idx + 1 < len(tokens):
            input_data = tokens[idx + 1]

    # Parse program arguments after --
    if "--" in tokens:
        idx = tokens.index("--")
        run_args = tokens[idx + 1:]

    # Write temp C++ file
    with tempfile.NamedTemporaryFile(suffix=".cpp", delete=False, mode="w") as tmp_cpp:
        tmp_cpp.write(cell)
        cpp_path = tmp_cpp.name
    exe_path = cpp_path[:-4] + ".exe"

    try:
        # Compile
        compile_proc = subprocess.run(
            ["g++", "-std=c++23", "-O2", "-Wall", cpp_path, "-o", exe_path],
            capture_output=True,
            text=True
        )
        if compile_proc.returncode != 0:
            print("❌ Compilation failed:\n", compile_proc.stderr)
            return

        # Run program
        run_proc = subprocess.run(
            [exe_path] + run_args,
            input=input_data,      # feed stdin here
            capture_output=True,
            text=True
        )
        if run_proc.stdout:
            print(run_proc.stdout, end="")
        if run_proc.stderr:
            print("⚠️ Runtime error:\n", run_proc.stderr)

    finally:
        for f in (cpp_path, exe_path):
            try: os.remove(f)
            except: pass
```

- ### Modification

**push back**

```python
%%cpp
#include <iostream>
#include <string>
using namespace std;

int main()
{
    string s;
    int n = 26;
    for (int i = 0; i < n; i++)
    {
        s.push_back('a' + i); // ASCII values
    }
    for (int i = 0; i < n; i++)
    {
        cout << s[i] << " ";
    }
}
```

**pop back**

```python
%%cpp
#include <iostream>
#include <string>
using namespace std;

int main()
{
    int n = 10;
    string s(n, '*');
    for (int i = 0; i < n; i++)
    {
        for (auto c : s)
        {
            cout << c << " ";
        }
        cout << endl;
        s.pop_back();
    }
}
```

**insert**

```python
%%cpp
#include <iostream>
#include <string>
using namespace std;

int main()
{
    int n = 5;
    string s(n, '-');
    for (int i = 0; i < 2*n; ++(++i))
    {
        
        s.insert(i, "+");
    }
    for (auto c : s)
        {
            cout << c << " ";
        }
        cout << endl;
}
```

**append**

```python
%%cpp
#include <iostream>
#include <string>
using namespace std;

int main()
{
    string s1 = "Hello, ";
    string s2 = "world";
    string s3 = "!";
    s1 += s2;
    cout << s1 << endl;
    s1.append(s3);
    cout << s1 << endl;
}
```

**erase**

```python
%%cpp
#include <iostream>
#include <string>
using namespace std;

int main()
{
    string s("Hhhi there!");
    s.erase(1, 2); // remove 2 characters starting from index 1
    for (auto c : s)
        {
            cout << c;
        }
        cout << endl;
}
```

**replacement**

```python
%%cpp
#include <iostream>
#include <string>
using namespace std;

int main()
{
    int i = 0;
    string s("Hi there!");
    s.replace(0, 2, "Hello,"); // start at index 0, replace 2 characters
    while (s[i])
    {
        cout << s[i];
        i++;
    }
    cout << endl;
}
```

- ### Conversion

**string to integer/float/double**

```python
%%cpp
#include <iostream>
#include <string>
using namespace std;

int main() {
    string s1 = "123";
    string s2 = "3.14";
    string s3 = "2.71";
    int num1 = stoi(s1); // string to integer
    float num2 = stod(s2); // string to float
    double num3 = stof(s3); // string to double
    cout << num1 << " " << num2 << " " << num3 << endl;
}
```

**integer/float/double to string**

```python
%%cpp
#include <iostream>
#include <string>
using namespace std;

int main() {
    int num1 = 123;
    float num2 = 3.14;
    double num3 = 2.71;
    string s1 = to_string(num1);
    string s2 = to_string(num2);
    string s3 = to_string(num3);
    cout << s1 << " " << s2 << " " << s3 << endl;
}
```

**ASCII to string**

```python
%%cpp
#include <iostream>
#include <string>
#include <string>
using namespace std;

int main() {
    string<int> ascii = {97, 98, 99, 100, 101};
    string result; 
    for (auto x : ascii)
    {
        result.push_back(char(x));
    }
    for (auto c : result)
    {
        cout << c << " ";
    }
    cout << endl;
}
```

- ### Function parameters

**reference (modifiable)**

```python
%%cpp
#include <iostream>
#include <string>
using namespace std;

void upper(string &s)
    {
        for (auto &c : s) 
        {
            if (c >= 97 && c <= 122) 
            {
                c -= 32;
            };
        }

    }

int main() 
{
    string s = "hello, WORLD!";
    upper(s);
    for (auto &c : s) 
        {
            cout << c;
        }
    cout << endl;
}
```

**const reference (read only)**

```python
%%cpp
#include <iostream>
#include <string>
using namespace std;

int cnt(const string &s, const char c)
{
    int sum = 0;
    for (auto x : s)
    {
        if (x == c)
        {
            sum++;
        }
    }
    return sum;
}

int main() {
   string s = "cbcsa";
   char c = 'c';
    cout << "Number of " << c << " in " << s << " is " << cnt(s, c) << endl;
}
```

**value (copied)**

```python
%%cpp
#include <iostream>
#include <string>
using namespace std;

string take_pure_char(string &s)
    {
        string result;
        for (auto &c : s) 
        {
            if (c >= 97 && c <= 122 || c >= 65 && c <= 90) 
            {
                result.push_back(c);
            };
        }
        return result;
    }

int main() 
{
    string s = "hello213world!";
    for (auto &c : take_pure_char(s)) 
        {
            cout << c;
        }
    cout << endl;
}
```
